```python
def prop(dfx):
    return dfx[["bill",'pizzas']].sum() / df[["bill","pizzas"]].sum()
    
driver_group.apply(prop)
```

```python
def prop(dfx, cols):
    return dfx[cols].sum() / df[cols].sum()
    
driver_group.apply(lambda dfx: prop(dfx, ["time","bill","free_wine"])).round(2)
```

```python
def meanGroupBy(df,group, cols):
    groups = df.groupby(group)
    return groups.apply(lambda dfx : dfx[cols].mean())
```

```python
meanGroupBy(df,"operator",'time')
```

```python
def meanGroupBy(df,group, cols):
    groups = df.groupby(group)
    return groups.apply(lambda dfx : dfx[cols].agg("mean"))
```

```python
meanGroupBy(df,"driver",'bill')
```
