```python
import pandas as pd
import numpy as np

# Set the random seed for reproducibility
np.random.seed(42)

# Create a DataFrame with random values
df = pd.DataFrame(np.random.randn(10, 5), columns=['A', 'B', 'C', 'D', 'E'])

# Calculate the number of null values needed
num_nulls = int(df.size * 0.4)

# Randomly choose positions to set as null
null_positions = np.random.choice(df.size, num_nulls, replace=False)

# Set the chosen positions to null
df.values.ravel()[null_positions] = np.nan
df.round(1)
```

```python
df.fillna(-100).round(1)
```

```python
df.fillna(method="bfill").round(1)
```

**fill by series**

```python
pd.Series([0,10,100,1000,10000], index=["A","B","C","D","E"])
```

```python
fill_series = pd.Series([0,10,100,1000,10000], index=["A","B","C","D","E"])
df.fillna(fill_series).round(1)
```

```python
df.fillna({"A":100,"B":200}).round(1)
```

```python
fill_mean_series =  df.mean().round(1)
fill_mean_series
```

```python
df.fillna(fill_mean_series).round(1)
```
