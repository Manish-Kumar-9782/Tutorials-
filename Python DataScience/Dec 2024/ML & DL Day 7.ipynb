```python
def meanGroupBy(df,group, cols):
    groups = df.groupby(group)
    return groups.apply(lambda dfx : dfx[cols].mean())
```

```python
meanGroupBy(df,"operator",'time')
```

```python
def meanGroupBy(df,group, cols):
    groups = df.groupby(group)
    return groups.apply(lambda dfx : dfx[cols].agg("mean"))
```

```python
meanGroupBy(df,"driver",'bill')
```

**aggregate group by function**

```python
def aggGroupBy(df,groupName, cols, aggfunc):
    groups = df.groupby(groupName)
    return groups.apply(lambda dfx: dfx[cols].agg(aggfunc))
```

```python
aggGroupBy(df, "operator", ["time","bill"],["sum","mean"])
```

**Null Replacing**

```python
# Set the random seed for reproducibility
np.random.seed(42)

# Create a DataFrame with random values
df_na = pd.DataFrame(np.random.randn(10, 5), columns=['A', 'B', 'C', 'D', 'E'])

# Calculate the number of null values needed
num_nulls = int(df_na.size * 0.4)

# Randomly choose positions to set as null
null_positions = np.random.choice(df_na.size, num_nulls, replace=False)

# Set the chosen positions to null
df_na.values.ravel()[null_positions] = np.nan
df_na.round(1)
```

**replace method**

```python
df_na.replace(np.nan, 100).round(1)
```
