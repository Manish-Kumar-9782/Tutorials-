```python
def talk(greetfunc, name, age):
    greetfunc()
    print(f"Hello my name is {name} and i am {age} yeras old")
    
```

```python
# func1
def talk(greetfunc):
    greetfunc()
    print(f"hello my name is Avi and i am 10 yeras old")
def greet(name="user"):
    print("hello", name, "how are you.....")
def outerfunc(func1, func2):
    print("call from outer function ")
    x=lambda:func1(func2)
    print(f" call of {func2.__name__} function has been wrapped by {func1.__name__} function..")
    return x
```

```python
outerfunc(talk, greet)
```

```python
talk(greet)
```
