Basic Example

```Python

# Define a list
my_list = [4, 7, 0]

# Create an iterator from the list
iterator = iter(my_list)

# Get elements from the iterator
print(next(iterator))  # Output: 4
print(next(iterator))  # Output: 7
print(next(iterator))  # Output: 0
```

In this example, we create an iterator from a list using the iter() function and retrieve elements one by one using the next() function.

Using a For Loop
A more elegant way to iterate through an iterator is by using a for loop:

```Python

# Define a list
my_list = [4, 7, 0]

# Iterate through the list using a for loop
for element in my_list:
    print(element)

```

You can also create your own custom iterator by defining a class with __iter__() and __next__() methods:

```Python

class PowTwo:
    """Class to implement an iterator of powers of two"""
    def __init__(self, max=0):
        self.max = max

    def __iter__(self):
        self.n = 0
        return self

    def __next__(self):
        if self.n <= self.max:
            result = 2 ** self.n
            self.n += 1
            return result
        else:
            raise StopIteration

# Create an object
numbers = PowTwo(3)

# Create an iterator from the object
i = iter(numbers)

# Iterate through the iterator
for num in i:
    print(num)

```

# Iterator class

````python
class PowTwo:
    """Class to implement an iterator of powers of two"""
    
    def __init__(self, max=0):
        self.max = max

    def __iter__(self):
        print("PowTwo Iterator initalized..")
        self.n = 0
        return self

    def __next__(self):
        print(f"iteration: {self.n+1}")
        if self.n <= self.max:
            result = 2 ** self.n
            self.n += 1
            return result
        else:
            raise StopIteration

    
````

````python
x =  PowTwo(5)
x_iter = iter(x)
````

````python
x_iter
````

````python
next(x_iter)
````

````python
for x in PowTwo(5):
    print(x)
````

````python
class UserIterator:

    def __init__(self):
        self.url = "https://jsonplaceholder.typicode.com/users"
        self.res = requests.get(self.url)

    def __iter__(self):
        if self.res.status_code == 200:  # if every thing is ok then go ahead
            self.data = self.res.json() # get data from response as json object like dict.
            self.max = len(self.data)  # number of users present inside the json list.
            self.n = 0 # initial 
            return self
        else:
            raise ValueError("No Data found to process")

    def __next__(self):

        if self.n < self.max:
            data = self.__extract__(self.n)
            self.n += 1
            return data
        else:
            raise StopIteration

    def __extract__(self, n):
        user = self.data[n]

        data = {"id":user["id"], "name": user["name"], "username":user["username"]}
        return data
        
````
