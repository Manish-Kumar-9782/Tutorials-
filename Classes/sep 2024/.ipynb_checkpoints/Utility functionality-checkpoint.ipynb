````python
rand_items = [random.randint(1,30) for i in range(1,20)]
````

````python
rand_items
````

````python
[num for num in rand_items if num%3==0 and  num % 2 == 0]
````

````python
for num in rand_items:
    if num % 3 == 0:
        print(num)
    else:
        continue
````

````python
genexp =  (num for num in rand_items if num%3==0 and  num % 2 == 0)
````

````python
genexp
````

````python
next(genexp)
````

**zip function**

````python
st = ["Aarav Sharma", "Vivaan Patel", "Aditya Singh", "Vihaan Gupta", "Arjun Kumar",
    "Sai Reddy"]
age = [random.randint(18,30) for i in st] 
````

````python
st,age
````

````python
# for student, age  in zip():
#     pass
````

````python
list(zip(st,age))
````

````python
for name, age  in zip(st,age):
    print(f"Student Name: {name:15s} \t age:{age:3d}")
````

**enumerate**

````python
for index, name in enumerate(st):
    print(f"Student Name: {name:15s} \t index:{index:3d}")
````

**Enum**

````python
from enum import Enum
````

````python
class TaskPriority(Enum):
    HIGH = "high"
    NORMAL = "normal"
    LOW = "low"
    CRITICAL = 'critical'
    REQUIRED = 'required'
````

````python
TaskPriority.REQUIRED
````

````python
class Days(Enum):
    SUNDAY = "Sunday"
    MONDAY = 'Monday'
    TUESDAY = "Tuesday"
    
````

````python
isinstance(Days.SUNDAY, Days)
````
