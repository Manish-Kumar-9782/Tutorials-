```python
from abc import ABC, abstractmethod
```

## Base Abstract Mobile Class

```python
class Mobile(ABC):

    @abstractmethod
    def call(self):
        pass

    @abstractmethod
    def message(self):
        pass
```

### Mobiles

```python
class SamsungMobile(Mobile):

    def __init__(self):

        self.phonebook = []

    def call(self):
        print("Samsung mobile calling.....")

    def message(self):
        print("Samsung mobile messaging to someone")
```

```python
class MotoMobile(Mobile):

    def __init__(self):
        self.phonebook = []

    def call(self):
        print("Moto mobile calling.....")

    def message(self):
        print("Moto mobile messaging to someone")
```

### Adding a constraint

```python
class Person:

    def __init__(self, name, mobile_phone=None):
        self.name = name
        
        if not None and isinstance(mobile_phone, Mobile):
            self.phone = mobile_phone
        elif not None and not isinstance(mobile_phone,Mobile):
            # raising an error of type of TypeError
            raise TypeError(f"mobile phone is not an instance of Mobile Type. given type {type(mobile_phone)}")
        else:
            self.phone = None
    

    def purchase_phone(self, mobile):
        self.phone = mobile

    def make_call(self):
        self.phone.call()

    def phone_call(self, mobile):
        mobile.call()
```

```python
# m1 
m1 = 10
pr = Person("Manish" , m1)
```

```python
m1 = SamsungMobile()
pr = Person("Manish" , m1)
```

```python
pr.purchase_phone("samsung mobile")
```

```python
pr.make_call()
```

## Adding flexibility for phone attribute

```python
def isMobile(obj):
    if isinstance(obj, Mobile):
        return True
    else:
        return False
        
def checkMobile(obj):
    if not isMobile(obj):
        raise TypeError(f"mobile phone is not an instance of Mobile Type.\
        given type {type(obj)}")
    

class Person:

    def __init__(self, name, mobile_phone=None):
        self.name = name
        
        # first check mobile
        checkMobile(mobile_phone) # if not an error 
        self.phone = mobile_phone

    def purchase_phone(self, mobile):
        checkMobile(mobile)
        self.phone = mobile

    def make_call(self):
        self.phone.call()

    def phone_call(self, mobile):
        if not isMobile(mobile):
            raise TypeError(f"Can not make a call from a type of {type(mobile)}'s object'. You Need a Mobile Type Object.")
        mobile.call()
```

```python
m1 = SamsungMobile()
pr = Person("Manish" , m1)
```

```python
pr.purchase_phone("phone")
```

```python
pr.phone_call("hello")
```

```python
pr.phone_call(m1)
```

```python
def isMobile(obj):
    if isinstance(obj, Mobile):
        return True
    else:
        return False
        
def checkMobile(obj):
    if not isMobile(obj):
        raise TypeError(f"mobile phone is not an instance of Mobile Type.\
        given type {type(obj)}")
    

class Person:

    def __init__(self, name, mobile_phone=None):
        self.name = name
        self.setPhone(mobile_phone)

    def getPhone(self):
        if isMobile(self._phone):
            return self._phone 
        else:
            errorMsg = f"MobileError: Can not reterieve mobile phone!. \
You have invalid object assigned for mobile phone. Invalid type {type(self._phone)} \
Please assign correct mobile phone instance"
            raise TypeError(errorMsg)

    def setPhone(self,phone):
        if phone is None:
            self._phone = None
            return
        
        if isMobile(phone):
            self._phone = phone
        else:
            raise TypeError("MobileError: assigned item is not instance of mobile.")

    def purchase_phone(self, mobile):
        self.setPhone(mobile)

    def make_call(self):
        # self.phone.call()
        # self._phone.call()
        self.getPhone().call()

    def phone_call(self, mobile):
        if not isMobile(mobile):
            raise TypeError(f"Can not make a call from a type of {type(mobile)}'s object'. You Need a Mobile Type Object.")
        mobile.call()
```
