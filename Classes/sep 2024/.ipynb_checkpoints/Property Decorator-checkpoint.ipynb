# Property Decorator

```python
from abc import ABC, abstractmethod
```

```python
class Mobile(ABC):

    @abstractmethod
    def call(self):
        pass

    @abstractmethod
    def message(self):
        pass
```

```python
class SamsungMobile(Mobile):

    def __init__(self):

        self.phonebook = []

    def call(self):
        print("Samsung mobile calling.....")

    def message(self):
        print("Samsung mobile messaging to someone")
```

```python
class MotoMobile(Mobile):

    def __init__(self):
        self.phonebook = []

    def call(self):
        print("Moto mobile calling.....")

    def message(self):
        print("Moto mobile messaging to someone")
```

```python
class Person:

    def __init__(self, name):
        self.name = name

    def setName(self, value):
        print("setName method running..")
        if type(value) != str:
            raise TypeError("Value must be a type of string")
        self._name = value

    def getName(self):
        print("getName method running..")
        return self._name
        
    name = property(fget=getName, fset=setName)
```

```python
pr = Person("rahul")
```

```python
pr.name # accessing data value
```

```python
pr.name = 123
```

### Property as decorator

```python
class Person:

    def __init__(self, name):
        self.name = name

    @property  # getter method of property.
    def name(self):
        print("getName method running..")
        return self._name

    @name.setter
    def name(self, value):
        print("setName method running..")
        if type(value) != str:
            raise TypeError("Value must be a type of string")
        self._name = value
```
