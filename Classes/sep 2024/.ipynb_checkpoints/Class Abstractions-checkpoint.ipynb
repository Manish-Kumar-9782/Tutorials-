```python
from abc import abstractmethod, ABC
```

## Abstract class

#### Stage 1

```python
class Mobile(ABC):

    @abstractmethod
    def call(self):
        pass

    @abstractmethod
    def message(self):
        pass
```

```python
# creating an object of Mobile
mobile = Mobile()
```

**Samsung Mobile class**

```python
class SamsungMobile(Mobile):

    def __init__(self):

        self.phonebook = []
        
```

```python
s_mobile = SamsungMobile()
```

### Stage 2
**implementing call & message method defination**

```python
class SamsungMobile(Mobile):

    def __init__(self):

        self.phonebook = []

    def call(self):
        print("Samsung mobile calling.....")

    def message(self):
        print("Samsung mobile messaging to someone")
```

```python
class MotoMobile(Mobile):

    def __init__(self):
        self.phonebook = []

    def call(self):
        print("Moto mobile calling.....")

    def message(self):
        print("Moto mobile messaging to someone")
```

```python
s_mobile = SamsungMobile()
```

```python
s_mobile.call()
```

```python
s_mobile.message()
```

```python
m_mobile = MotoMobile()
```

```python
m_mobile.call()
```

```python
m_mobile.message()
```
