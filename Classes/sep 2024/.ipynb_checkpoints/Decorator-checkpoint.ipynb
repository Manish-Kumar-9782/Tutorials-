**Higher Order Functions**

A higher-order function is a function that either takes one or more functions as arguments or returns a function as its result. This concept is fundamental in functional programming and allows for more abstract and flexible code. Python supports higher-order functions, enabling developers to create more reusable and modular code.

```python
def talk(greetfunc, name, age):
    greetfunc()
    print(f"Hello my name is {name} and i am {age} year's old..")
```

```python
## func1
def talk(greetfunc):
    greetfunc()
    print(f"Hello my name is Rahul and i am 20 year's old..")

## func2
def greet(name="user"):
    print("Hello", name, "how are you..")

def outerfunc(func1, func2):

    print("call from outer function")
    x = lambda : func1(func2)
    print(f"call of {func2.__name__} function has been wrapped by {func1.__name__} function...")

    return x
    
```

```python
x =   outerfunc(talk, greet)
```

```python
x()
```

```python
def fullname(fname, lname):
    return f"{fname} {lname}"
```

```python
fullname("manish", "kumar")
```

```python
def modify_fullname(func):

        def fullname_wrapper(fname, lname):
            # before calling statement
            result = func(fname, lname)
            # after calling statement
            result = result.center(50, "*")
            result = result.upper()
            
            return result
            
        return fullname_wrapper
```

```python
# if original function defination can't be defined by us then we will use this syntax.
fullname_wrapper = modify_fullname(fullname)
```

```python
fullname_wrapper
```

```python
fullname_wrapper("manish", "kumar")
```

```python
fullname("manish", "kumar")
```

**using generic names for creating decoratorss**

```python
def debug(func):

        def wrapper(fname, lname):
            func_name = func.__name__
            print(f"Calling {func_name} function {func_name}({fname}, {lname})")
            # before calling statement
            result = func(fname, lname)
            # after calling statement
            print(f"function call has finished and function {func_name} returned with ''{result}''")
            
            return result
            
        return wrapper
```

```python
@modify_fullname
def fullname(fname, lname):
    return f"{fname} {lname}"

fullname("manish", "kumar")
```

**Time Decorator**

```python
import time
import requests
```

```python
def evaltime(func):

    def wrapper(*args, **kwargs):
        # befor func call
        start_time = time.time()
        result = func(*args, **kwargs)
        # after func call
        end_time = time.time()
        print(f"\nfunction {func.__name__}'s total run time: {(end_time-start_time):.2f}\n", )
        return result

    return wrapper
        
```

```python
@evaltime
def get_data():
    fetch_url = "https://jsonplaceholder.typicode.com/users"
    return requests.get(fetch_url)

get_data()
```

```python
@evaltime
def run():
    for i in range(100):
        time.sleep(0.0001)
        print(i,end=' ')

run()
```

## Authentication

```python
class User:

    def __init__(self,name, email, isLogged=False, isRegistered=False):
        self.name = name
        self.email = email
        self.isLoggedIn = isLogged
        self.isAuthenticated = isRegistered

class AuthentcationError(Exception):
    pass
```

```python
user_1 = User("Manish", "mansih123@gmail.com", isRegistered=True)
user_x = User("Anonymoms", "jamesbond00x@gmail.com")
```

```python
def authenticate(func):
    def wrapper(user, *args, **kwargs):
        # befor func call
        
        if not isinstance(user, User):
            raise ValueError("User Required...")

        if not user.isAuthenticated:
            raise AuthentcationError("User is not Authenticated...")
            
        result = func(user, *args, **kwargs)
        # after func call
        return result

    return wrapper
```

```python
@authenticate
def get_data(user):
    fetch_url = "https://jsonplaceholder.typicode.com/users"
    return requests.get(fetch_url)

get_data(user_x)
```

**Function Routing**

```python
ROUTES = {}

def route(route):

    def router(func):
        
        def wrapper(*args, **kwargs):
            # befor func call    
            result = func(*args, **kwargs)
            # after func call
            return result
            
        ROUTES[route] = wrapper
        return wrapper

    return router
```

```python
@route("/hello")
def say_hello():
    print("Hello world")

@route("/")
def home_page():
    print("Welcome to home page")

@route("/user")
def show_user(user):
    print("Name: ", user.name)
    print("Email: ", user.email)
    print("LoggedIn: ", user.isLoggedIn)
    print("Registered: ", user.isAuthenticated)
```

```python
def fetch_url(route, **func_kwargs):
    if route in ROUTES.keys():
        func = ROUTES[route]
        func(**func_kwargs)
    else:
        raise ValueError(f"No route exist with this for route:{route}", )
```
