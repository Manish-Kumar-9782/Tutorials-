## classes Inheritance

```python
class A:

    def __init__(self):
    self.attr1 = 1
    self.attr2 = 2

    def method1(self):
        pass

# inheriting A with B
class B(A):

    def __init__(self):
        super().__init__()
        self.attr3 = 3
        self.attr4 = 4

    def method4(self):
        pass
```

````python
class Mobile:

    def __init__(self):
        self.phonebook = []

    def call(self):
        print("mobile calling")
````

````python
class SamsungMobile(Mobile):

    def __init__(self):
        super().__init__()  
        # Note: if we don't use super() method to call __init__ method of Class Mobile
        # then we can't access attribute of Mobile class in SamsungMobile class.
        self.brand = 'samsung'
        self.model = 'Galaxy A1'

    def camara(self):
        print("camera opened....")
        print("clicked a photo...")
````

````python
mobile = Mobile()
````

````python
mobile.phonebook
````

````python
mobile.call()
````

**creating a Sumsung mobile object**

````python
s_mobile = SamsungMobile() 
````

````python
s_mobile.phonebook
````

````python
s_mobile.call()
````

````python
s_mobile.camara()
````

### Adding more methods to Mobile

````python
class Mobile:

    def __init__(self):
        self.phonebook = []

    def call(self):
        print("mobile calling")

    def message(self):
        print("Sendig a message")

    def audio(self):
        print("playing audio")

    def video(self):
        print("playing video")
````

````python
class SamsungMobile(Mobile):

    def __init__(self):
        super().__init__()  
        # Note: if we don't use super() method to call __init__ method of Class Mobile
        # then we can't access attribute of Mobile class in SamsungMobile class.
        self.brand = 'samsung'
        self.model = 'Galaxy A1'

    def camara(self):
        print("camera opened....")
        print("clicked a photo...")
````

````python
mobile = Mobile()
````

````python
mobile.audio()
````

````python
s_mobile = SamsungMobile()
````

````python
s_mobile.audio()
````

## Overriding call method

````python
class SamsungMobile(Mobile):

    def __init__(self):
        super().__init__()  
        # Note: if we don't use super() method to call __init__ method of Class Mobile
        # then we can't access attribute of Mobile class in SamsungMobile class.
        self.brand = 'samsung'
        self.model = 'Galaxy A1'

    def camara(self):
        print("camera opened....")
        print("clicked a photo...")

    def call(self):
        print("samsung mobile calling to someone")
````

````python
mobile = Mobile()
````

````python
mobile.call()
````

````python
s_mobile = SamsungMobile()
````

````python
s_mobile.call()
````

## Overriding example 2

````python
class Mobile:

    def __init__(self):
        self.phonebook = []

    def call(self):
        print("mobile calling")
````

````python
class SamsungMobile(Mobile):

    def __init__(self):
        super().__init__()  
        # Note: if we don't use super() method to call __init__ method of Class Mobile
        # then we can't access attribute of Mobile class in SamsungMobile class.
        self.brand = 'samsung'
        self.model = 'Galaxy A1'

    def camara(self):
        print("camera opened....")
        print("clicked a photo...")

    def call(self):
        super().call()
        print("samsung mobile calling to someone")
````

````python
s_mobile = SamsungMobile()
````

````python
s_mobile.call()
````
