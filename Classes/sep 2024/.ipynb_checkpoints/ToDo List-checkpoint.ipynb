```python
import os
```

```python
os.path.join("files", "mylist")
```

```python
class TodoList:

    def __init__(self, task, type="ol"):
        self.base_path = "files" 
        self.task = task
        self._taskType = type
        self.path = os.path.join(self.base_path, f"{task}.txt")
        self._tasks = []
    
    def addTask(self,task):
        if isinstance(task, str):
            self._tasks.append(task)
        else:
            raise TypeError("task must be an instance of string.")


    def addTasks(self,tasks):
        if isinstance(tasks, (list, set, tuple)):
            for task in tasks:
                self.addTask(task)
        else:
            raise TypeError("tasks must sequence type object like list,set or tuple.")

    def save(self):
        file = open(self.path, 'w')
        file.writelines((task+'\n' for task in self._tasks))
        file.close()

    def display(self):
        print("Task: ", self.task)
        print("=========================================")

        for i, task in enumerate(self._tasks):
            if self._taskType == "ol":
                print(f"{i+1:3.0f} {task}")
            elif self._taskType == "ul":
                print("* ", task)
            else:
                raise TypeError("Todo List only supports ol (ordered list) and  ul (unordered list) operation")
    
    def updateTask(self,index, task):

        if  index < len(self._tasks):
            
            if isinstance(task, str):
                self._tasks[index] = task
            else:
                raise TypeError("task must be an instance of string.")
        else:
            print(f"Warning: index number {index} does not exist" )
```

```python
todo =  TodoList("test-2", "ul")

todo.addTask("go to market")

todo.addTask("go to home")

todo.addTasks(["This is item one", "This is item two", "This is item thre"])

todo.save()
```

```python
todo.display()
```

```python
todo.updateTask(3, "Make a cup of tea")
```
