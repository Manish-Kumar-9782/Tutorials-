**Problem 1:** Write a Python script to generate and print a dictionary that contains a number (between 1 and n) in the form (x, x*x).

Sample Dictionary ( n = 5) :

Expected Output : {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

```python
for x in range(1,10):
    print(x)
```

```python
# now create an empty dictionary 
mydict = dict()  # or {}
for x in range(1,10):
    # create a key as value of x
    mydict[x] = x*x
print(mydict)
```

**using n number to change the series.**

```python
# now create an empty dictionary 
mydict = dict()  # or {}
n = 50 # storing our n values as integer
for x in range(1,n):
    # create a key as value of x
    mydict[x] = x*x
print(mydict)
```

```python
def generate_square_series(n):
    mydict = dict() # or {}
    for x in range(1,n+1):
        mydict[x] = x*x

    return mydict
```

```python
generate_square_series(10)
```

```python
n = int(input("Enter a number"))
generate_square_series(n)
```

**Problem: 2** Write a Python program to print a specified list after removing the 0th, 4th and 5th elements.

Sample List : ['Red', 'Green', 'White', 'Black', 'Pink', 'Yellow']

Expected Output : ['Green', 'White', 'Black']

```python
delete_indexes = [0,4,5]

data_list = ['Red', 'Green', 'White', 'Black', 'Pink', 'Yellow']
n = len(data_list)

new_list = []
for i in range(n):
    if i not in delete_indexes:
        new_list.append(data_list[i])
```

***Write a Python program to generate and print a list of the first and last 5 elements where the values are square numbers between 1 and 30 (both included).**

```python
def printValues():
    l = list()
    for i in range(1, 100):  
        l.append(i**2)
    print(l[:5])

    print(l[-1:])

printValues()
```
