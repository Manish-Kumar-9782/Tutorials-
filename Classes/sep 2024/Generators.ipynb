````python
def gen_qubes(start,end):
    for x in gen_squares(start,end):
        yield x**3
        
````

````python
def add_randomness():
    for x in gen_qubes(1,10):
        yield x+random.randint(10000000,999999999)
````

````python
for x in add_randomness():
    print(x)
````

````python
def rand_sqrt():
    for x in add_randomness():
        yield math.sqrt(x)
````

````python
def removeSpace():
    # this email varaible contains raw email address
    for email in email_list:
        yield email.strip()
# =================================== #
def removeDomain():
    # this email address contains trimed/striped email address
    for email in removeSpace():
        if email.find("@") > 0:
            splitted_email = email.split("@")
            yield splitted_email[0]

# =================================== #
def extractNames():
    for string in removeDomain():
        splitted_string = string.split(".")
        capitalized_names = []     
        for x_str in splitted_string:
            capitalized_names.append(x_str.capitalize())
        yield capitalized_names
# =================================== #

def extractNamesEfficient():
    for string in removeDomain():
        splitted_string = string.split(".")

        def cap(_str):
            return _str.capitalize()
        yield map(cap, splitted_string)
````

## DataPipeline

````python
class DataPipeline:

    def __init__(self):
        self.pipes = []

    def addPipe(self,pipe):

        if callable(pipe):
            self.pipes.append(pipe)
        else:
            current_pine_number = len(self.pipes)
            raise TypeError(f"pipe {current_pine_number + 1} is not callable generator")
        

    def run(self, initialData):
        data = initialData
        for pipe in self.pipes:
            data = pipe(data)  # here pipe will return a generator
            
        return list(data)
````

````python
def removeSpace(data):
    # this email varaible contains raw email address
    for item in data:
        yield item.strip()
# =================================== #
def removeDomain(data):
    # this email address contains trimed/striped email address
    for item in data:
        if item.find("@") > 0:
            item_splitted = item.split("@")
            yield item_splitted[0]

def extractNames(data):
    for string in data:
        splitted_string = string.split(".")
        capitalized_names = []     
        for x_str in splitted_string:
            capitalized_names.append(x_str.capitalize())
        yield tuple(capitalized_names)


def joinText(data):
    for textSequence in data:
        yield "-".join(textSequence)
````

````python
x = 10

try:
    pipeline = DataPipeline()
    pipeline.addPipe(removeSpace)
    pipeline.addPipe(removeDomain)
    pipeline.addPipe(extractNames)
    pipeline.addPipe(joinText)
    pipeline.addPipe(x)
    
    pipeline.run(email_list)
except TypeError as e:
    print(e)
except Exception:
    print("Unknown Error Found: ", e)
````
