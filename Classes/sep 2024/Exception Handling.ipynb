```python
from abc import ABC, abstractmethod
```

## Base Abstract Mobile Class

```python
class Mobile(ABC):

    @abstractmethod
    def call(self):
        pass

    @abstractmethod
    def message(self):
        pass
```

### Mobiles

```python
class SamsungMobile(Mobile):

    def __init__(self):

        self.phonebook = []

    def call(self):
        print("Samsung mobile calling.....")

    def message(self):
        print("Samsung mobile messaging to someone")
```

```python
class MotoMobile(Mobile):

    def __init__(self):
        self.phonebook = []

    def call(self):
        print("Moto mobile calling.....")

    def message(self):
        print("Moto mobile messaging to someone")
```

### Adding a constraint

```python
class Person:

    def __init__(self, name, mobile_phone=None):
        self.name = name
        
        if not None and isinstance(mobile_phone, Mobile):
            self.phone = mobile_phone
        elif not None and not isinstance(mobile_phone,Mobile):
            # raising an error of type of TypeError
            raise TypeError(f"mobile phone is not an instance of Mobile Type. given type {type(mobile_phone)}")
        else:
            self.phone = None
    

    def purchase_phone(self, mobile):
        self.phone = mobile

    def make_call(self):
        self.phone.call()

    def phone_call(self, mobile):
        mobile.call()
```

## Try except block

```python
try:
    s1 = SamsungMobile()
    m1 = MotoMobile()
    pr = Person("Rahul")

    pr.make_call() #  No Mobile Error

    pr.phone_call()

except Exception as e:
    print(e)


print("program ended..")
```

```python
class MobileNotFound(Exception):
    pass

class InvalidMobileType(Exception):
    pass
```

```python
class Person:

    def __init__(self, name, mobile_phone=None):
        self.name = name
        self.setPhone(mobile_phone)

    def getPhone(self):
        if isMobile(self._phone):
            return self._phone 
        else:
            errorMsg = f"Can not reterieve mobile phone!. \
You have invalid object assigned for mobile phone. Invalid type {type(self._phone)} \
Please assign correct mobile phone instance"
            raise MobileNotFound(errorMsg)

    def setPhone(self,phone):
        if phone is None:
            self._phone = None
            return
        
        if isMobile(phone):
            self._phone = phone
        else:
            raise InvalidMobileType("Assigned item is not instance of mobile.")

    def purchase_phone(self, mobile):
        self.setPhone(mobile)

    def make_call(self):
        # self.phone.call()
        # self._phone.call()
        self.getPhone().call()

    def phone_call(self, mobile):
        if not isMobile(mobile):
            raise InvalidMobileType(f"Can not make a call from a type of {type(mobile)}'s object'. You Need a Mobile Type Object.")
        mobile.call()
```

```python
try:
    s1 = SamsungMobile()
    m1 = MotoMobile()
    pr = Person("Rahul", s1) # may get InvalidMobileType

    pr.make_call()  # may get MobileNotFound Exception

    pr.phone_call(m1)
    10/0  # Error : division by zero

except InvalidMobileType as e:
    print(e)
    print("Please provide a valid mobile object, you can have it by using any Mobile sub-class.")
    
except MobileNotFound as e:
    # print(e)
    print("You don't have any mobile available to make a call.")

except Exception as ex:
    print("Unknown error found: ", ex)

print("\n---------------------------------")
print("program ended..")
```
